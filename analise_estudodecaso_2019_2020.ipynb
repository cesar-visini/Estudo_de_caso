Importação de bibliotecas necessárias - pandas

```python
import pandas as pd # importando a biblioteca pandas para manipulação de dados
```

Download de data frames com os arquivos no formato csv de 2019 e 2020

```python
q1_2019 = pd.read_csv(r'C:\Users\cesar\Projetos Python\Estudo de caso 1\Divvy_Trips_2019_Q1 - Divvy_Trips_2019_Q1.csv') #df 2019 bruto
q1_2020 = pd.read_csv(r'C:\Users\cesar\Projetos Python\Estudo de caso 1\Divvy_Trips_2020_Q1 - Divvy_Trips_2020_Q1.csv') #df 2020 bruto
```

Inspeção das colunas dos dataframes e junção dos dois dataframes em apenas um.

```python
# checagem das colunas dos dataframes
print(f'Columns in q1_2019:\n{q1_2019.columns}\n')
print(f'Columns in q1_2020:\n{q1_2020.columns}\n')
```

```python
# renomeando as colunas do dataframe q1_2019 para que fiquem iguais às do dataframe q1_2020
q1_2019 = q1_2019.rename(columns={
    'trip_id': 'ride_id',
    'bikeid': 'rideable_type',
    'start_time': 'started_at',
    'end_time': 'ended_at',
    'from_station_name': 'start_station_name',
    'from_station_id': 'start_station_id',
    'to_station_name': 'end_station_name',
    'to_station_id': 'end_station_id',
    'usertype': 'member_casual'
    })
print()

# checagem das informações dos dataframes
print(f'q1_2019 data infos:')
q1_2019.info()
print(f'q1_2020 data infos:')
q1_2020.info()
```

```python
# Ajustando tipos de dados (q1_2019): ride_id para str, rideable_type para str 
q1_2019['ride_id'] = q1_2019['ride_id'].astype(str)
q1_2019['rideable_type'] = q1_2019['rideable_type'].astype(str)
```

```python
# Concatenando os dois data frames: q1_2019 e q1_2020
all_trips = pd.concat([q1_2019, q1_2020], ignore_index = True)
```

```python
# Remoção dos dados de latitude e longitude, aniversário e gênero
all_trips = all_trips.drop(columns = ['start_lat','start_lng','end_lat','end_lng', 'birthyear', 'gender', 'tripduration'], errors = 'ignore')
```

Limpeza e manutenção dos dados para preparar para a analise

```python
# inspeção das informações do dataframe all_trips
print(f'\nLista do nome das colunas\n{all_trips.columns}\n')
print(f'\nQuantas linhas tem o dataframe? {len(all_trips)}')
print(f'\nQuais são das dimensões do dataframe? {all_trips.shape}')
print(f'\nQuais são as 6 primeiras linhas do dataframe?\n{all_trips.head(6)}')
print(f'\nLista de tipos de dados das colunas do dataframe:\n{all_trips.dtypes}')
print(f'\nValores únicos na coluna "member_casual": {all_trips["member_casual"].unique()}')
all_trips.info()

# inspeção dos dados estatísticos do dataframe
print("\nDados estatísticos do dataframe:\n", all_trips.describe())
```

Problema 1: Reescrever Subscriber como Member e Customer como Casual

```python
# limpeza da coluna member_casual, substituindo os valores "Subscriber" por "member" e "Customer" por "casual"
print(f'\nValores únicos na coluna "member_casual": {all_trips["member_casual"].unique()}')
all_trips['member_casual'] = all_trips['member_casual'].replace({
'Subscriber': 'member',
'Customer': 'casual'
})

# checagem dos valores únicos na coluna member_casual após a limpeza
print("\nValue counts for 'member_casual' column after cleaning:\n", all_trips['member_casual'].value_counts())
```

Problema 2: Converter started_at e ended_at em formato de tempo, criar uma coluna e calcular a duração de cada viagem

```python
# convertendo as colunas started_at e ended_at para o tipo datetime
all_trips['started_at'] = pd.to_datetime(all_trips['started_at'])
all_trips['ended_at'] = pd.to_datetime(all_trips['ended_at'])

# criando novas colunas para análise: date, month, day, year, day_of_week e ride_length
all_trips['date'] = all_trips['started_at'].dt.date
all_trips['month'] = all_trips['started_at'].dt.month_name()
all_trips['day'] = all_trips['started_at'].dt.day
all_trips['year'] = all_trips['started_at'].dt.year
all_trips['day_of_week'] = all_trips['started_at'].dt.day_name()
all_trips['ride_length'] = (all_trips['ended_at'] - all_trips['started_at']).dt.total_seconds() / 60 # convertendo o tempo de corrida para minutos

# checagem dos tipos de dados das colunas do dataframe all_trips após a criação da coluna ride_length
print("\nTipos de dados depois da criação da coluna ride_length:\n")
all_trips.info()

```

Problema 3: Removendo dados sujos dos casos em que as bicicletas são retiradas de fora das docas para a realização de inspeções e manutenções.

```python
# removendo viagens com tempo negativo e viagens que começam fora da estação HQ QR
all_trips_v2 = all_trips[(all_trips['start_station_name'] != 'HQ QR') & (all_trips['ride_length'] >= 0)].copy() 
all_trips_v2['month'] = all_trips_v2['started_at'].dt.month_name()
```

Analise estatística dos dados do data frame all_trips_v2

```python
 # Exibindo a descrição estatística da coluna ride_length
d1 = (
    all_trips_v2["ride_length"]
    .describe()
)
print(f'Descrição estatística da coluna ride_length:\n{d1}\n')

# Comparando a coluna ride_length entre membros e clientes casuais
d2 = (
    all_trips_v2
    .groupby("member_casual")["ride_length"]
    .describe()
)
print(f'Comparação da coluna ride_length entre membros e clientes casuais:\n{d2}\n') 

# Comparando a coluna ride_length entre membros e clientes casuais nos dias da semana
d3 = (
    all_trips_v2
    .groupby(["member_casual", "day_of_week"])["ride_length"]
    .mean()
)
print(f'Comparação da coluna ride_length entre membros e clientes casuais nos dias da semana:\n{d3}\n') 

# Colocando os dias da semana em ordem correta para análise
days_order = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
all_trips_v2['day_of_week'] = pd.Categorical(all_trips_v2['day_of_week'], categories=days_order, ordered=True)

# Analisando a média de duração das viagens por dia da semana para membros e clientes casuais com os dias da semana em ordem correta
d4 = (
    all_trips_v2
    .groupby(["member_casual", "day_of_week"])["ride_length"]
    .mean()
)
print(f'Comparação da coluna ride_length entre membros e clientes casuais nos dias da semana:\n{d4}\n')

# Analisando o número de usuários por tipo de assinatura (membro ou casual) e por dia da semana
d5 = (
    all_trips_v2
    .groupby(["member_casual", "day_of_week"])
    .agg(
        number_of_rides=('ride_id', 'count'),
        average_duration=('ride_length', 'mean')
    )
    .reset_index()
)
print(f'\nNúmero de usuários por tipo de assinatura (membro ou casual) e por dia da semana:\n{d5}\n')
```
